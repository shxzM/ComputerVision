```python
import os
import numpy as np
import tensorflow as tf
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras.models import Model
from tensorflow.keras.layers import (
    Input, Conv2D, BatchNormalization, Activation, Add, Dense,
    GlobalAveragePooling2D, Dropout, Multiply, Reshape, DepthwiseConv2D
)
import matplotlib.pyplot as plt
from google.colab import drive
import zipfile
from sklearn.metrics import precision_recall_curve, auc

# Enable mixed precision training
tf.keras.mixed_precision.set_global_policy('mixed_float16')

def extract_dataset():
    zip_path = "/content/drive/MyDrive/Copy of PLANT_VILLAGE_SPLIT.zip"
    extract_path = "/content/plant_village"

    if not os.path.exists(extract_path):
        print("Extracting dataset...")
        with zipfile.ZipFile(zip_path, 'r') as zip_ref:
            zip_ref.extractall(extract_path)
        print("Extraction complete!")
    else:
        print("Dataset already extracted!")
    return extract_path

def create_activation(activation='swish'):
    if activation == 'swish':
        return tf.keras.activations.swish
    return Activation(activation)

class SEBlock:
    def __init__(self, filters, ratio=4):
        self.filters = filters
        self.ratio = ratio

    def __call__(self, x):
        squeeze = GlobalAveragePooling2D()(x)
        excitation = Dense(self.filters // self.ratio, activation='swish')(squeeze)
        excitation = Dense(self.filters, activation='sigmoid')(excitation)
        excitation = Reshape((1, 1, self.filters))(excitation)
        return Multiply()([x, excitation])

def MBConv(x, filters, kernel_size, expansion_ratio, stride=1, se_ratio=4, dropout_rate=0.2):
    input_filters = x.shape[-1]
    expanded_filters = input_filters * expansion_ratio

    # Expansion phase
    if expansion_ratio != 1:
        expand = Conv2D(expanded_filters, 1, padding='same', use_bias=False)(x)
        expand = BatchNormalization()(expand)
        expand = create_activation('swish')(expand)
    else:
        expand = x

    # Depthwise phase
    depthwise = DepthwiseConv2D(kernel_size, stride, padding='same', use_bias=False)(expand)
    depthwise = BatchNormalization()(depthwise)
    depthwise = create_activation('swish')(depthwise)

    # SE block
    if se_ratio:
        depthwise = SEBlock(expanded_filters, se_ratio)(depthwise)

    # Output phase
    output = Conv2D(filters, 1, padding='same', use_bias=False)(depthwise)
    output = BatchNormalization()(output)

    # Residual connection
    if stride == 1 and input_filters == filters:
        if dropout_rate:
            output = Dropout(dropout_rate)(output)
        output = Add()([x, output])

    return output

def FusedMBConv(x, filters, kernel_size, expansion_ratio, stride=1, se_ratio=4, dropout_rate=0.2):
    input_filters = x.shape[-1]
    expanded_filters = input_filters * expansion_ratio

    # Fused expansion and depthwise phase
    if expansion_ratio != 1:
        fused = Conv2D(expanded_filters, kernel_size, stride, padding='same', use_bias=False)(x)
        fused = BatchNormalization()(fused)
        fused = create_activation('swish')(fused)
    else:
        fused = x

    # SE block
    if se_ratio:
        fused = SEBlock(expanded_filters, se_ratio)(fused)

    # Output phase
    output = Conv2D(filters, 1, padding='same', use_bias=False)(fused)
    output = BatchNormalization()(output)

    # Residual connection
    if stride == 1 and input_filters == filters:
        if dropout_rate:
            output = Dropout(dropout_rate)(output)
        output = Add()([x, output])

    return output

def create_efficientnetv2(input_size=(128, 128, 3), num_classes=38):
    inputs = Input(input_size)

    # Initial convolution
    x = Conv2D(24, 3, strides=2, padding='same', use_bias=False)(inputs)
    x = BatchNormalization()(x)
    x = create_activation('swish')(x)

    # EfficientNetV2 stages
    # Stage 1: Fused-MBConv
    x = FusedMBConv(x, 24, 3, expansion_ratio=1, stride=1)
    x = FusedMBConv(x, 24, 3, expansion_ratio=1, stride=1)

    # Stage 2: Fused-MBConv
    x = FusedMBConv(x, 48, 3, expansion_ratio=4, stride=2)
    x = FusedMBConv(x, 48, 3, expansion_ratio=4, stride=1)

    # Stage 3: Fused-MBConv
    x = FusedMBConv(x, 80, 3, expansion_ratio=4, stride=2)
    x = FusedMBConv(x, 80, 3, expansion_ratio=4, stride=1)

    # Stage 4: MBConv
    x = MBConv(x, 160, 3, expansion_ratio=4, stride=2)
    x = MBConv(x, 160, 3, expansion_ratio=4, stride=1)
    x = MBConv(x, 160, 3, expansion_ratio=4, stride=1)

    # Stage 5: MBConv
    x = MBConv(x, 256, 3, expansion_ratio=6, stride=2)
    x = MBConv(x, 256, 3, expansion_ratio=6, stride=1)
    x = MBConv(x, 256, 3, expansion_ratio=6, stride=1)

    # Head
    x = Conv2D(1280, 1, padding='same', use_bias=False)(x)
    x = BatchNormalization()(x)
    x = create_activation('swish')(x)

    # Classification head
    x = GlobalAveragePooling2D()(x)
    x = Dropout(0.5)(x)
    outputs = Dense(num_classes, activation='softmax')(x)

    model = Model(inputs=inputs, outputs=outputs)
    return model

def load_data(base_path, img_size=(128, 128)):
    data_generator = ImageDataGenerator(
        rescale=1./255,
        horizontal_flip=True,
        rotation_range=20,
        zoom_range=0.15,
        width_shift_range=0.2,
        height_shift_range=0.2,
        shear_range=0.15,
        fill_mode="nearest"
    )

    train_generator = data_generator.flow_from_directory(
        os.path.join(base_path, 'PLANT_VILLAGE_SPLIT', 'train'),
        target_size=img_size,
        batch_size=32,
        class_mode='categorical',
        shuffle=True
    )

    val_generator = ImageDataGenerator(rescale=1./255).flow_from_directory(
        os.path.join(base_path, 'PLANT_VILLAGE_SPLIT', 'val'),
        target_size=img_size,
        batch_size=32,
        class_mode='categorical',
        shuffle=False
    )

    test_generator = ImageDataGenerator(rescale=1./255).flow_from_directory(
        os.path.join(base_path, 'PLANT_VILLAGE_SPLIT', 'test'),
        target_size=img_size,
        batch_size=32,
        class_mode='categorical',
        shuffle=False
    )

    print(f"\nTotal training samples: {train_generator.n}")
    print(f"Total validation samples: {val_generator.n}")
    print(f"Total test samples: {test_generator.n}")
    print(f"Steps per epoch: {len(train_generator)}")

    return train_generator, val_generator, test_generator

def plot_results(history, model, test_generator, num_classes):
    plt.figure(figsize=(12, 4))

    plt.subplot(1, 2, 1)
    plt.plot(history.history['accuracy'], label='Training Accuracy')
    plt.plot(history.history['val_accuracy'], label='Validation Accuracy')
    plt.title('Model Accuracy')
    plt.xlabel('Epoch')
    plt.ylabel('Accuracy')
    plt.legend()

    plt.subplot(1, 2, 2)
    plt.plot(history.history['loss'], label='Training Loss')
    plt.plot(history.history['val_loss'], label='Validation Loss')
    plt.title('Model Loss')
    plt.xlabel('Epoch')
    plt.ylabel('Loss')
    plt.legend()

    plt.tight_layout()
    plt.show()

    y_pred = model.predict(test_generator)
    y_test = test_generator.labels
    y_test = tf.keras.utils.to_categorical(y_test, num_classes=num_classes)

    plt.figure(figsize=(8, 6))
    precision = dict()
    recall = dict()

    for i in range(num_classes):
        precision[i], recall[i], _ = precision_recall_curve(y_test[:, i], y_pred[:, i])

    precision["micro"], recall["micro"], _ = precision_recall_curve(y_test.ravel(), y_pred.ravel())

    plt.plot(recall["micro"], precision["micro"],
             label=f'Micro-averaged PR curve (area = {auc(recall["micro"], precision["micro"]):.2f})')
    plt.xlabel('Recall')
    plt.ylabel('Precision')
    plt.title('Precision-Recall Curve')
    plt.legend()
    plt.show()

def train_and_evaluate():
    base_path = extract_dataset()
    print("Loading data...")
    train_generator, val_generator, test_generator = load_data(base_path)
    num_classes = len(train_generator.class_indices)

    print("Creating EfficientNetV2 model...")
    model = create_efficientnetv2(num_classes=num_classes)

    optimizer = tf.keras.optimizers.AdamW(
        learning_rate=0.001,
        weight_decay=0.0001
    )

    model.compile(
        optimizer=optimizer,
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    callbacks = [
        tf.keras.callbacks.EarlyStopping(
            monitor='val_loss',
            patience=5,
            restore_best_weights=True
        ),
        tf.keras.callbacks.ReduceLROnPlateau(
            monitor='val_loss',
            factor=0.2,
            patience=2,
            min_lr=1e-6
        ),
        tf.keras.callbacks.ModelCheckpoint(
            'best_efficientnetv2_model.h5',
            monitor='val_accuracy',
            save_best_only=True,
            mode='max'
        )
    ]

    print("Starting training...")
    history = model.fit(
        train_generator,
        epochs=20,
        validation_data=val_generator,
        callbacks=callbacks,
        verbose=1
    )

    # Load best weights
    model.load_weights('best_efficientnetv2_model.h5')

    # Evaluate and plot results
    test_loss, test_accuracy = model.evaluate(test_generator)
    plot_results(history, model, test_generator, num_classes)

    return (history.history['accuracy'][-1],
            history.history['val_accuracy'][-1],
            test_accuracy)

if __name__ == "__main__":
    # Mount Google Drive at the start
    drive.mount('/content/drive')

    print("Starting the plant disease classification process using EfficientNetV2...")
    train_acc, val_acc, test_acc = train_and_evaluate()

    print(f"\nFinal Accuracies:")
    print(f"Training Accuracy: {train_acc:.4f}")
    print(f"Validation Accuracy: {val_acc:.4f}")
    print(f"Testing Accuracy: {test_acc:.4f}")
```
